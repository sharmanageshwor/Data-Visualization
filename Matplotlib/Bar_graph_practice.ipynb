#**Bar Graph Practice**

```python
import matplotlib.pyplot as plt
```

##Simple Bar chat

```python
x=[1,2,3,4]
y=[2,4,6,8]

c=['r','g','y','b']
plt.xlabel("Xmen",fontsize=20,color="red")
plt.ylabel("no of mens",fontsize=20)
plt.title("Marvel",fontsize=30)
plt.bar(x,y,color=c,alpha=0.5,width=0.4,edgecolor="red")
plt.show()
```

##Bar Graph with legend

```python
x=['Ram','Shyam','Hari','Gopal']
y=[34,45,56,67]

c=['r','g','m','b']
plt.xlabel("Name of students",fontsize=18)
plt.ylabel("Marks",fontsize=18)
plt.title("Marks of students",fontsize=22)
plt.bar(x,y,width=0.4,color=c,edgecolor='black',linewidth=2,linestyle='--',alpha=0.8,label='marks')
plt.legend()
plt.show()
```

##Bar Graph with multiple Data

```python
x=['Ram','Shyam','Hari','Gopal']
y=[34,45,56,67]
z=[20,30,46,40]
c=['r','g','m','b']
plt.xlabel("Name of students",fontsize=18)
plt.ylabel("Marks",fontsize=18)
plt.title("Marks of students",fontsize=22)
plt.bar(x,y,width=0.4,color='r',label='marks')
plt.bar(x,z,width=0.4,color='y',label='marks2')
plt.legend()
plt.show()
```

#Represent Multiple Bar side by side

```python
import matplotlib.pyplot as plt
import numpy as np


x=['Ram','Shyam','Hari','Gopal']
y=[34,45,56,67]
z=[20,30,46,40]
width=0.4

p=np.arange(len(x))
p1=[j+width for j in p]
c=['r','g','m','b']

plt.xlabel("Name of students",fontsize=18)
plt.ylabel("Marks",fontsize=18)
plt.title("Marks of students",fontsize=22)

plt.bar(p,y,width,color='r',label='marks')
plt.bar(p1,z,width,color='y',label='marks2')
plt.xticks(p+width/2,x,rotation=20)
plt.legend()
plt.show()
```

##Showing Bar Chat in Horizontal view

```python
import matplotlib.pyplot as plt


x=['Ram','Shyam','Hari','Gopal']
y=[34,45,56,67]
z=[20,30,46,40]

plt.xlabel("Name of students",fontsize=18)
plt.ylabel("Marks",fontsize=18)
plt.title("Marks of students",fontsize=22)

plt.barh(x,y,color='r',label='marks')
plt.barh(x,z,color='y',label='marks2')

plt.legend()
plt.show()
```
