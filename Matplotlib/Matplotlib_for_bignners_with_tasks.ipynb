## 1. What is Matplotlib?

Matplotlib turns numbers into pictures — line charts, bar charts, histograms, scatter plots, pie charts.

**Why learn it?**
- It is the base library that most other Python chart tools are built on
- Works in notebooks, scripts, and websites
- Full control over how a chart looks

**When to use it vs other tools:**
| Situation | Best Tool |
|---|---|
| Quick chart, full control | Matplotlib |
| Statistical chart, less code | Seaborn |
| Chart straight from a table of data | Pandas `.plot()` |
| Interactive chart (hover, zoom) | Plotly / Bokeh |

## 3. Your First Chart — a Simple Line

Two steps: give some numbers, then tell matplotlib to draw them.

````python
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

plt.plot(x, y)   # draw a line through the points
plt.show()        # display it
````

**What happened:** `plt.plot(x, y)` connects each `(x, y)` pair with a line. `plt.show()` displays the picture.

## 4. Adding a Title and Axis Labels

Same chart as before — we just add 3 new lines.

````python
plt.plot(x, y)
plt.title("Sales Over 5 Days")   # NEW: title on top
plt.xlabel("Day")                 # NEW: label under x-axis
plt.ylabel("Units Sold")          # NEW: label beside y-axis
plt.show()
````

# Matplotlib Line Styles and Markers

In Matplotlib, we can change the **line style** and **marker style** using `linestyle` and `marker`.

## 1. Line Styles

| Linestyle | Meaning |
|---|---|
| `'-'` | Solid line |
| `'--'` | Dashed line |
| `'-.'` | Dash-dot line |
| `':'` | Dotted line |
| `'None'` | No line |

## 2. Marker Styles

| Marker | Meaning |
|---|---|
| `'o'` | Circle |
| `'s'` | Square |
| `'^'` | Triangle up |
| `'v'` | Triangle down |
| `'*'` | Star |
| `'+'` | Plus |
| `'x'` | X |
| `'D'` | Diamond |
| `'.'` | Point |

## 3. Example

```python
import matplotlib.pyplot as plt

x = ["Mon", "Tue", "Wed", "Thu", "Fri"]
y = [20, 35, 30, 45, 40]

         color='green',

**What changed:** same line, but now it is green, dashed, and shows a dot at every data point.

## 6. Two Lines on the Same Chart

Just call `plt.plot()` a second time before `plt.show()`.

````python
product_a = [10, 20, 15, 30, 25]
product_b = [5, 15, 25, 10, 20]

plt.plot(x, product_a, marker='o', label='Product A')
plt.plot(x, product_b, marker='s', label='Product B')   # NEW: second line
plt.title("Product Sales Comparison")
plt.xlabel("Day")
plt.ylabel("Units Sold")
plt.show()
````

## 8. Pie Chart

Use `plt.pie()` to show parts of a whole (use only 2-6 slices).

````python
labels = ['Rent', 'Food', 'Transport', 'Savings']
expenses = [40, 25, 15, 20]

plt.pie(expenses, labels=labels, autopct='%1.1f%%')
plt.title("Monthly Budget")
plt.show()
````

Actually, the important part is:

% → start of formatting
1.1 → display 1 decimal place
f → floating-point number
%% → print the actual % symbol

````python
studnet_exp = ['Rent', 'Food', 'cloths', 'books']
expenses_amount = [40.2, 25.4, 15.5, 20.3]

plt.pie(expenses_amount, labels=studnet_exp, autopct='%1.1f%%')
plt.title("Monthly Budget")
plt.show()
````

## 9. Histogram — Built Step by Step

A histogram shows how data is **spread out** (how many values fall in each range).

We will build it slowly. Each step below adds **one new thing** to the same chart, so you can see exactly what changed.

### Step 1: Just the data

First, create some sample data — 1000 random height values.

````python
import matplotlib.pyplot as plt

marks = [35, 38, 42, 43, 44, 46, 48, 49, 50,
         51, 53, 54, 56, 58, 60, 62, 64]

bins = [30,35, 40,45, 50,55, 60,65, 70]

plt.hist(marks, bins=bins,
         edgecolor='black')

plt.title("Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

plt.show()
````

````python
import matplotlib.pyplot as plt

marks = [35, 38, 42, 43, 44, 46, 48, 49, 50,
         51, 53, 54, 56, 58, 60, 62, 64]

# Define class intervals
bins = [30, 35, 40, 45, 50, 55, 60, 65, 70]

# Create histogram
plt.hist(marks,
         bins=bins,
         color='skyblue',
         edgecolor='black',
         alpha=0.7)

# Add title and labels
plt.title("Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

# Add grid
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.show()
````

### Step 2: Draw the most basic histogram

Call `plt.hist()` on the data. No settings yet — just the default.

**What changed:** the bars are now orange with a black border, making individual bars easier to tell apart.

**What changed:** the chart now clearly says what it is showing — heights on the x-axis, how many people had that height on the y-axis.

## 10. Scatter Plot — Built Step by Step

A scatter plot shows the relationship between **two** numeric variables — one dot per data pair.

We will build it the same way as the histogram: one small change at a time.

### Step 1: Just the data

Create two lists of the same length — hours studied and the exam score for 8 students.

````python
hours_studied = [1, 2, 3, 4, 5, 6, 7, 8]
exam_score =    [50, 55, 65, 60, 70, 75, 85, 90]

print(list(zip(hours_studied, exam_score)))   # peek at the pairs
````

**What we have:** 8 pairs of numbers — student 1 studied 1 hour and scored 50, student 2 studied 2 hours and scored 55, and so on. Nothing plotted yet.

### Step 2: Draw the most basic scatter plot

Call `plt.scatter(x, y)`.

````python
plt.scatter(hours_studied, exam_score)
plt.show()
````

**What changed:** each pair of numbers is now a single dot on the chart — no lines connecting them, just points.

### Step 3: Add color

Add `color` to change the dot color from the default blue.

````python
plt.scatter(hours_studied, exam_score, color='purple')   # NEW: purple dots
plt.show()
````

**What changed:** same dots, just recolored purple.

### Step 4: Change dot size

Add `s` (size) to make the dots bigger or smaller.

````python
plt.scatter(hours_studied, exam_score, color='purple', s=100)   # NEW: s=100 makes dots bigger
plt.show()
````

**What changed:** the dots are now larger and easier to see. Try `s=20` for smaller dots.

### Step 5: Add title and axis labels

Same as always — make it readable.

````python
plt.scatter(hours_studied, exam_score, color='purple', s=100)
plt.title("Study Hours vs Exam Score")   # NEW
plt.xlabel("Hours Studied")               # NEW
plt.ylabel("Exam Score")                  # NEW
plt.show()
````

**What changed:** now anyone looking at the chart knows exactly what it's measuring.

````python
hours_studied = [1, 2, 3, 4, 5, 6, 7, 8]
exam_score =    [50, 55, 65, 60, 70, 75, 85, 90]

````

### Step 6: Color dots by a third value

Instead of one flat color, pass a **list of numbers** to `c=` — matplotlib will color each dot based on its value, like a heat scale.

````python
plt.scatter(hours_studied, exam_score,
            c=exam_score,        # NEW: color each dot by its exam score
            s=20,
            cmap='viridis')      # NEW: cmap = the color scale to use
plt.title("Study Hours vs Exam Score")
plt.xlabel("Hours Studied")
plt.ylabel("Exam Score")
plt.colorbar(label='Exam Score')   # NEW: shows what the colors mean
plt.show()
````

While changing the

**What changed:** instead of every dot being the same purple, low scores are now a different color than high scores, and a colorbar explains the scale.

### Step 7 (bonus): Compare two groups on one scatter plot

Just like the histogram, add a second `plt.scatter()` call to compare two groups.

````python
group_a_hours = [1, 2, 3, 4, 5]
group_a_scores = [50, 55, 60, 58, 65]

group_b_hours = [4, 5, 6, 7, 8]
group_b_scores = [70, 75, 80, 85, 90]

plt.scatter(group_a_hours, group_a_scores, color='blue', s=100, label='Class A')
plt.scatter(group_b_hours, group_b_scores, color='red', s=100, label='Class B')   # NEW second group
plt.title("Study Hours vs Exam Score: Class A vs Class B")
plt.xlabel("Hours Studied")
plt.ylabel("Exam Score")
plt.legend()   # NEW
plt.show()
````

**What changed:** a second `plt.scatter()` call was layered on the same chart, with its own color and label, so we can compare the two classes directly.

## 13. Alternative Libraries — and When to Use Them

| Library | Best For | Value |
|---|---|---|
| **Matplotlib** | Full control, static charts | Most flexible, works everywhere |
| **Seaborn** | Statistical plots (heatmaps, box plots) | Prettier defaults, less code |
| **Pandas `.plot()`** | Quick chart from a table of data | Fastest for data you already loaded |
| **Plotly** | Interactive charts, dashboards | Hover, zoom, web-ready |
| **Bokeh** | Big data, interactive dashboards | Handles large datasets well |
| **Altair** | Short, declarative chart code | Very readable syntax |

**Simple rule:** exploring data → Pandas. Statistics → Seaborn. Need interactivity → Plotly/Bokeh. Need full control → Matplotlib.
